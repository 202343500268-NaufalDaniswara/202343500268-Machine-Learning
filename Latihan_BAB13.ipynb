```python
# Import library yang dibutuhkan
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
import time

print("TensorFlow version:", tf.__version__)
```

```python
# 1. Load Dataset CIFAR-10
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

# 2. Membuat Subset (misal: 5000 train, 1000 test)
# Normalisasi pixel values ke range 0-1
train_images, test_images = train_images / 255.0, test_images / 255.0

# Mengambil subset data
X_train_sub = train_images[:5000]
y_train_sub = train_labels[:5000]
X_test_sub = test_images[:1000]
y_test_sub = test_labels[:1000]

print(f"Train shape: {X_train_sub.shape}")
print(f"Test shape: {X_test_sub.shape}")

# 3. Membangun Model CNN sesuai Modul
model_cnn = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 4. Compile Model
model_cnn.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

model_cnn.summary()

# 5. Latih Model (Catat waktu dan jumlah parameter)
start_time_cnn = time.time()

history_cnn = model_cnn.fit(X_train_sub, y_train_sub, epochs=10,
                            validation_data=(X_test_sub, y_test_sub))

end_time_cnn = time.time()
training_time_cnn = end_time_cnn - start_time_cnn

# Evaluasi
test_loss_cnn, test_acc_cnn = model_cnn.evaluate(X_test_sub, y_test_sub, verbose=2)
params_cnn = model_cnn.count_params()

print(f"\n--- HASIL CNN FROM SCRATCH ---")
print(f"Akurasi Test: {test_acc_cnn*100:.2f}%")
print(f"Waktu Pelatihan: {training_time_cnn:.2f} detik")
print(f"Jumlah Parameter: {params_cnn}")
```

```python
import time
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# ============================================================
# PRAKTIKUM 2 (FIXED): Transfer Learning MobileNetV2
# ============================================================

IMG_SIZE = 96  # ukuran target, harus salah satu dari [96, 128, 160, 192, 224]

# 1. Load base model — input_shape HARUS sama dengan hasil resize
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False  # freeze base model

# 2. Bangun model lengkap
model_tl = models.Sequential([
    layers.Resizing(IMG_SIZE, IMG_SIZE),                       # 32x32 -> 96x96
    layers.Lambda(tf.keras.applications.mobilenet_v2.preprocess_input),
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 3. Compile
model_tl.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# PENTING: panggil build() agar layer Dense terbentuk sebelum summary
model_tl.build(input_shape=(None, 32, 32, 3))
model_tl.summary()

# 4. Training
start_time_tl = time.time()
history_tl = model_tl.fit(
    X_train_sub, y_train_sub,
    epochs=10,
    validation_data=(X_test_sub, y_test_sub)
)
training_time_tl = time.time() - start_time_tl

# 5. Evaluasi
test_loss_tl, test_acc_tl = model_tl.evaluate(X_test_sub, y_test_sub, verbose=2)
trainable_params_tl = np.sum([tf.keras.backend.count_params(w) for w in model_tl.trainable_weights])

print(f"\n--- HASIL TRANSFER LEARNING (MobileNetV2) ---")
print(f"Akurasi Test        : {test_acc_tl*100:.2f}%")
print(f"Waktu Pelatihan     : {training_time_tl:.2f} detik")
print(f"Parameter Trainable : {trainable_params_tl}")
```
